````python
import google.generativeai as genai
import os
from dotenv import load_dotenv
from openai import OpenAI
import json

dotenv_path = "../.env"
_ = load_dotenv(dotenv_path=dotenv_path)

api_key =os.getenv("GEMINI_API_KEY")
genai.configure(api_key=api_key)
# genai.configure(api_key=os.environ["API_KEY"])
````

````python
def ask_chatgpt(text: str, prompt_path=None):
    # Check if a prompt path is provided and read prompt text
    dotenv_path = "../.env"
    _ = load_dotenv(dotenv_path=dotenv_path)
    api_key =os.getenv("OPENAI_API_KEY")
    client = OpenAI(api_key = api_key)
    if prompt_path and text:
        with open(prompt_path, 'r') as file:
            prompt = file.read().strip()
    else:
        return "Error: no data given"

    # Model configuration - replace 'gpt-4' with the specific model if needed
    model_name = "gpt-4o-2024-11-20"  # or "gpt-3.5-turbo" if you want a different model
    # Send the prompt to the model
    response = client.chat.completions.create(
    model=model_name,
    messages=[
        {'role' : "system" , "content" : prompt },
        {'role': "user" , "content": text}
    ],
    temperature=0.1,
    top_p=0.2
    )
        
    # Return the response content
    return response.choices[0].message.content
````

````python
import json
import re

def convert_to_tables(data):
    # Get the "original" table directly.
    original_table = data.get("original", "")
    
    same_tables_text = data.get("same_tables", "")
    
    # First, try to extract tables that are enclosed within ```markdown ... ``` blocks.
    pattern_backticks = r"```markdown\n(.*?)\n```"
    tables = re.findall(pattern_backticks, same_tables_text, re.DOTALL)
    
    # If we don't get exactly 3 tables from the backticks method, try an alternate approach.
    if len(tables) != 3:
        pattern_headers = r"### Table \d+:[^\n]*\n\n(.*?)(?=\n\n---|\Z)"
        tables = re.findall(pattern_headers, same_tables_text, re.DOTALL)
    
    # Warn if we still don't have exactly 3 tables.
    if len(tables) != 3:
        print("Warning: Expected to find 3 markdown tables but found", len(tables))
    
    # Prepare the result dictionary with keys "original", "Table1", "Table2", "Table3".
    result = {
        "original": original_table,
        "Table1": tables[0] if len(tables) > 0 else "",
        "Table2": tables[1] if len(tables) > 1 else "",
        "Table3": tables[2] if len(tables) > 2 else ""
    }
    
    return result

# Sample JSON input string
input_json = r'''{
  "original": "| Distance (yards) | Greyhound | Time | Date |\n| ---------------- | --------- | ---- | ---- |\n| 325 | Lemon Clover | 17.34 | 11.10.1996 |\n| 525 | Whitty Guinness | 28.54 | 29.10.2010 |\n| 550 | Whatsupjack | 29.91 | 18.09.2009 |\n| 700 | Tinas Girl | 38.79 | 19.08.2003 |\n| 790 | Shining Rumble | 44.76 | 13.07.2004 |",
  "same_tables": "### Table 1: Transposed Table with Semantic Rephrasing\n```markdown\n| Attribute         | Entry 1         | Entry 2          | Entry 3       | Entry 4       | Entry 5          |\n| ----------------- | --------------- | ---------------- | ------------- | ------------- | ---------------- |\n| Distance (yds)    | 325             | 525              | 550           | 700           | 790              |\n| Dog Name          | Lemon Clover    | Whitty Guinness  | Whatsupjack   | Tinas Girl    | Shining Rumble   |\n| Duration (secs)   | 17.34           | 28.54            | 29.91         | 38.79         | 44.76            |\n| Event Date        | 11.10.1996      | 29.10.2010       | 18.09.2009    | 19.08.2003    | 13.07.2004       |\n```\n\n---\n\n### Table 2: Column Abbreviation and Unit Conversion (yards to meters)\n```markdown\n| Dist. (m) | Greyhound         | Time (s) | Date       |\n| --------- | ----------------- | -------- | ---------- |\n| 297.18    | Lemon Clover      | 17.34    | 11.10.1996 |\n| 480.06    | Whitty Guinness   | 28.54    | 29.10.2010 |\n| 502.92    | Whatsupjack       | 29.91    | 18.09.2009 |\n| 640.08    | Tinas Girl        | 38.79    | 19.08.2003 |\n| 722.43    | Shining Rumble    | 44.76    | 13.07.2004 |\n```\n\n---\n\n### Table 3: Row-Column Swap and Semantic Rephrasing\n```markdown\n| Greyhound         | Distance (yards) | Duration (seconds) | Event Date   |\n| ----------------- | ---------------- | ------------------ | ------------ |\n| Lemon Clover      | 325             | 17.34             | 11.10.1996   |\n| Whitty Guinness   | 525             | 28.54             | 29.10.2010   |\n| Whatsupjack       | 550             | 29.91             | 18.09.2009   |\n| Tinas Girl        | 700             | 38.79             | 19.08.2003   |\n| Shining Rumble    | 790             | 44.76             | 13.07.2004   |\n```"
}'''

# Parse the JSON string into a Python dictionary
data = json.loads(input_json)

# Convert the data into a dictionary with 3 tables and the original table.
output_tables = convert_to_tables(data)

# Print the result dictionary.
print(json.dumps(output_tables, indent=2))
````

````python
import json

# Example dataset: list of dictionaries.
# dataset = [
#     {
#         "original": "Original_keyValue1",
#         "Same1": "Same1_keyValue1",
#         "Same2": "Same2_keyValue1",
#         "Same3": "Same3_keyValue1",
#         "Diff1": "Diff1_keyValue1",
#         "Diff2": "Diff2_keyValue1",
#         "Diff3": "Diff3_keyValue1"
#     },
#     {
#         "original": "Original_keyValue2",
#         "Same1": "Same1_keyValue2",
#         "Same2": "Same2_keyValue2",
#         "Same3": "Same3_keyValue2",
#         "Diff1": "Diff1_keyValue2",
#         "Diff2": "Diff2_keyValue2",
#         "Diff3": "Diff3_keyValue2"
#     },
#     # Add more records as needed...
# ]
dataset = spec_data

# This dictionary will hold the continuous indexed combinations.
continuous_combinations = {}
counter = 1  # Global counter for continuous indexing

# Process each record in the dataset.
for record in dataset:
    original_value = record.get("original")
    if original_value is None:
        raise ValueError("A record is missing the 'original' key.")
    
    # Loop through the keys of the record, skipping the "original" key.
    for key, value in record.items():
        if key == "original":
            continue
        # Assign the combination to the continuous counter.
        continuous_combinations[str(counter)] = [original_value, value]
        counter += 1

# Write the resulting dictionary to a JSON file.
with open("continuous_combinations.json", "w") as outfile:
    json.dump(continuous_combinations, outfile, indent=4)

print("Continuous combinations have been saved to 'continuous_combinations.json'.")
````
