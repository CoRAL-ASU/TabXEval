```python
import numpy as np

def markdown_to_numpy_with_headers(markdown_table):
    # Split the markdown table into lines
    lines = markdown_table.strip().split('\n')
    
    # Extract the header and data
    header = [col.strip() for col in lines[0].split('|') if col.strip()]
    data = [
        [col.strip() for col in line.split('|') if col.strip()]
        for line in lines[2:]  # Skip the header and separator lines
    ]
    
    # Combine header and data into a single list
    combined_data = [header] + data
    
    # Convert to a NumPy array
    numpy_array = np.array(combined_data, dtype=object)
    
    return numpy_array

# Example usage:
markdown_table = '''| Name | Age | City |
|------|-----|------|
| John | 30  | New York |
| Jane | 25  | Los Angeles |'''

numpy_array_with_headers = markdown_to_numpy_with_headers(table1)
print(numpy_array_with_headers)
```

```python
import json

with open('../tables.json', 'r') as file:
    dataset = json.load(file)

for data in dataset:
    original = data.get("original", "")
    for i in range(5):  # Assuming there are up to 5 perturbations (0-4)
        perturbation_key = f"pertubation{i}"
        result_key = f"result{i}"
        if perturbation_key in data:
            result = evaluate_table(markdown_to_numpy_with_headers(data[perturbation_key]), markdown_to_numpy_with_headers(original),eval_exact_match=True)
            # alignment_result = ask_chatgpt(TableAlign(original, data[perturbation_key]), 'prompt.txt')  
            data[result_key] = result
            print(data[f"result{i}"])

# Write the updated dataset to a new JSON file
with open('em_over_syn.json', 'w') as file:
    json.dump(dataset, file, indent=2)

print("Updated JSON file 'output.json' has been created successfully.")
```
